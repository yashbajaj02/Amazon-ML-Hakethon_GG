```python
import os
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../dataset/train")

# Load sample of training data (TSV format)
s1_sample = pd.read_csv(DATA_DIR / "train_source1.tsv", sep="\t", nrows=1000)
s2_sample = pd.read_csv(DATA_DIR / "train_source2.tsv", sep="\t", nrows=1000)
s3_sample = pd.read_csv(DATA_DIR / "train_source3.tsv", sep="\t", nrows=1000)
gt_sample = pd.read_csv(DATA_DIR / "train_ground_truth.tsv", sep="\t", nrows=1000)

print("Source 1 sample shape:", s1_sample.shape)
print("Source 2 sample shape:", s2_sample.shape)
print("Source 3 sample shape:", s3_sample.shape)
print("Ground Truth sample shape:", gt_sample.shape)
```
